**1. WAP in python to implement a simple table with operations to insert symbols , search for a symbol & display the contains of the table**

```python
symbol_table = {}

def insert_symbol():
    name = input("Enter symbol name: ")
    dtype = input("Enter data type: ")
    address = input("Enter memory address: ")

    symbol_table[name] = {
        "Type": dtype,
        "Address": address
    }
    print("Symbol inserted successfully.\n")

def search_symbol():
    name = input("Enter symbol to search: ")

    if name in symbol_table:
        print("Symbol Found:")
        print(symbol_table[name])
    else:
        print("Symbol not found.")

def display_table():
    print("\n------ SYMBOL TABLE ------")
    print("{:<10} {:<10} {:<10}".format("Symbol", "Type", "Address"))
    for key, value in symbol_table.items():
        print("{:<10} {:<10} {:<10}".format(
            key,
            value["Type"],
            value["Address"]
        ))
    print()

while True:
    print("1.Insert")
    print("2.Search")
    print("3.Display")
    print("4.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        insert_symbol()
    elif choice == 2:
        search_symbol()
    elif choice == 3:
        display_table()
    elif choice == 4:
        break
```

**2. write a python program to implent all basic single table operations , including modifications and deletions.**

```python
symbol_table = {}

def insert_symbol():
    name = input("Enter symbol name: ")
    dtype = input("Enter data type: ")
    address = input("Enter memory address: ")

    symbol_table[name] = {
        "Type": dtype,
        "Address": address
    }
    print("Symbol inserted successfully.\n")

def search_symbol():
    name = input("Enter symbol to search: ")

    if name in symbol_table:
        print("Symbol Found:")
        print(symbol_table[name])
    else:
        print("Symbol not found.")

def display_table():
    print("\n------ SYMBOL TABLE ------")
    print("{:<10} {:<10} {:<10}".format("Symbol", "Type", "Address"))
    for key, value in symbol_table.items():
        print("{:<10} {:<10} {:<10}".format(
            key,
            value["Type"],
            value["Address"]
        ))
    print()

def modify_table():
    name = input("Enter symbol to modify: ")
    if name in symbol_table:
        symbol_table[name]["Type"] = input("Enter new data type: ")
        symbol_table[name]["Address"] = input("Enter new memory address: ")
        print("Symbol modified successfully.\n")
    else:
        print("Symbol not found.")

def delete_table():
    name = input("Enter symbol to delete: ")
    if name in symbol_table:
        del symbol_table[name]
        print("Symbol deleted successfully.\n")
    else:
        print("Symbol not found.")


while True:
    print("1.Insert")
    print("2.Search")
    print("3.Display")
    print("4.Modify")
    print("5.Delete")
    print("6.Break")

    choice = int(input("Enter choice: "))

    if choice == 1:
        insert_symbol()
    elif choice == 2:
        search_symbol()
    elif choice == 3:
        display_table()
    elif choice == 4:
        modify_table()

    elif choice == 5:
        delete_table()


    elif choice == 6:
        break
```
