**Write a Python Program to read assemble level and their address and generate a single table.**

```python
# Symbol Table for Assembly Code

symbol_table = {}

def generate_symbol_table(code):
    address = 100  # Starting address

    for line in code:
        parts = line.strip().split()

        # Check if line contains a label (ends with :)
        if parts[0].endswith(":"):
            label = parts[0][:-1]
            symbol_table[label] = address

        address += 4  # Assume each instruction occupies 4 bytes

    return symbol_table


# User Input
n = int(input("Enter the number of assembly instructions: "))

assembly_code = []

print("Enter the assembly instructions:")
for i in range(n):
    line = input()
    assembly_code.append(line)

# Generate Symbol Table
table = generate_symbol_table(assembly_code)

# Display Symbol Table
print("\nSymbol Table")
print("-----------------------")
print("Symbol\tAddress")
print("-----------------------")

for symbol, addr in table.items():
    print(f"{symbol}\t{addr}")
```

**WAP in python program to stimulate the location counter while processing assembly instruction .**

```python
# Python Program to Simulate Location Counter

def location_counter():
    start = int(input("Enter the starting address: "))
    n = int(input("Enter the number of assembly instructions: "))

    lc = start

    print("\nLocation Counter Table")
    print("-" * 50)
    print("{:<10}{}".format("Address", "Instruction"))
    print("-" * 50)

    for i in range(n):
        instruction = input(f"Instruction {i+1}: ")
        print("{:<10}{}".format(lc, instruction))
        lc += 4      # Assume each instruction is 4 bytes

    print("-" * 50)
    print("Final Location Counter =", lc)

# Call the function
location_counter()
```

**Write a python program to implement a symbol table with nested scopes (Global and Local ) for managing variables in different blocks.**

```python
# Python Program to Implement Symbol Table with Nested Scopes

global_scope = {}
local_scope = {}

# Global Scope
g = int(input("Enter number of Global variables: "))

print("\nEnter Global Variables:")
for i in range(g):
    var = input("Variable Name: ")
    value = input("Value: ")
    global_scope[var] = value

# Local Scope
l = int(input("\nEnter number of Local variables: "))

print("\nEnter Local Variables:")
for i in range(l):
    var = input("Variable Name: ")
    value = input("Value: ")
    local_scope[var] = value

# Display Symbol Table
print("\nGLOBAL SYMBOL TABLE")
print("---------------------------")
print("Variable\tValue")
for var, value in global_scope.items():
    print(f"{var}\t\t{value}")

print("\nLOCAL SYMBOL TABLE")
print("---------------------------")
print("Variable\tValue")
for var, value in local_scope.items():
    print(f"{var}\t\t{value}")

# Search Variable
search = input("\nEnter variable to search: ")

if search in local_scope:
    print(f"{search} found in LOCAL scope with value = {local_scope[search]}")
elif search in global_scope:
    print(f"{search} found in GLOBAL scope with value = {global_scope[search]}")
else:
    print("Variable not found.")
```

**write a python program to implement path 1 of a 2 path assembler by generating the symbol table and intermediate code.**

```python
locctr = int(input("Enter starting address: "))
n = int(input("Enter number of assembly statements: "))

symbol_table = {}
intermediate = []


print("\nEnter statements in the format:")
print("LABEL OPCODE OPERAND")
print("(Use - if LABEL is absent)\n")

for _ in range(n):
    parts = input().split()

    if len(parts) != 3:
        print("Invalid input format")
        continue

    label, opcode, operand = parts

    if label != "-":
        symbol_table[label] = locctr

    intermediate.append((locctr, opcode, operand))

    locctr += 4

print("\n----- SYMBOL TABLE -----")
print("Label\tAddress")

for label, addr in symbol_table.items():
    print(f"{label}\t{addr}")

print("\n----- INTERMEDIATE CODE -----")
print("Address\tOpcode\tOperand")

for item in intermediate:
    print(f"{item[0]}\t{item[1]}\t{item[2]}")
```
