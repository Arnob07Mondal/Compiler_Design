**Write a Python Program Lexical Analyser (Scanner) to identify keywords
,Operators,constant,delimiter and Spacial Symbol.**

```python
keywords = {
    "int", "float", "char", "double", "if", "else", "while", "for",
    "return", "break", "continue", "void", "main", "switch", "case",
    "default", "do", "struct", "typedef", "const", "unsigned", "long",
    "short", "signed", "sizeof"
}

# Operators
operators = {
    "+", "-", "*", "/", "%", "=",
    "==", "!=", "<", ">", "<=", ">=",
    "&&", "||", "!", "++", "--"
}

# Delimiters
delimiters = {
    ";", ",", "(", ")", "{", "}", "[", "]"
}

# Special Symbols
special_symbols = {
    "#", "$", "@", "&"
}

print("Enter a C program (Type END in a new line to finish):\n")

program = ""

while True:
    line = input()
    if line.upper() == "END":
        break
    program += line + " "

# Insert spaces around delimiters and operators
symbols = [';', ',', '(', ')', '{', '}', '[', ']',
           '+', '-', '*', '/', '%', '=', '<', '>', '!', '&', '|', '#', '$', '@']

for s in symbols:
    program = program.replace(s, ' ' + s + ' ')

tokens = program.split()

print("\n------------ LEXICAL ANALYZER OUTPUT ------------\n")

for token in tokens:

    # Keyword
    if token in keywords:
        print(token, " --> Keyword")

    # Operator
    elif token in operators:
        print(token, " --> Operator")

    # Delimiter
    elif token in delimiters:
        print(token, " --> Delimiter")

    # Special Symbol
    elif token in special_symbols:
        print(token, " --> Special Symbol")

    # Integer Constant
    elif token.isdigit():
        print(token, " --> Integer Constant")

    # Floating Constant
    elif token.replace('.', '', 1).isdigit() and token.count('.') == 1:
        print(token, " --> Floating Constant")

    # Identifier
    elif token[0].isalpha() or token[0] == '_':
        print(token, " --> Identifier")

    # Unknown Token
    else:
        print(token, " --> Invalid Token")
```

**Write a python program that counts number of keywords ,
identifier,operators,Constant,delimiters,literals in a given Source Program**

```python
import keyword
import re

# Input from user
code = input("Enter a Python statement:\n")

# Tokenize the input
tokens = re.findall(r'\w+|==|!=|<=|>=|[-+*/%=<>;:,.(){}\[\]]|".*?"|\'.*?\'', code)

# Sets
operator_set = {
    "+", "-", "*", "/", "%", "=",
    "==", "!=", "<", ">", "<=", ">="
}

delimiter_set = {
    "(", ")", "{", "}", "[", "]",
    ",", ";", ":", "."
}

# Lists
keywords = []
identifiers = []
operators = []
constants = []
delimiters = []
literals = []

# Classify tokens
for token in tokens:
    if keyword.iskeyword(token):
        keywords.append(token)
    elif token in operator_set:
        operators.append(token)
    elif token in delimiter_set:
        delimiters.append(token)
    elif token.isdigit():
        constants.append(token)
    elif (
        (token.startswith('"') and token.endswith('"')) or
        (token.startswith("'") and token.endswith("'"))
    ):
        literals.append(token)
    elif token.isidentifier():
        identifiers.append(token)

# Display Result
print("\n----- Result -----")
print("Keywords   :", len(keywords))
print("Identifiers:", len(identifiers))
print("Operators  :", len(operators))
print("Constants  :", len(constants))
print("Delimiters :", len(delimiters))
print("Literals   :", len(literals))

print("\nDetails")
print("Keywords   :", keywords)
print("Identifiers:", identifiers)
print("Operators  :", operators)
print("Constants  :", constants)
print("Delimiters :", delimiters)
print("Literals   :", literals)
```

**Write a Python Program to removal comments (single line and Multi Line ) from a Source
program.**

```python
print("Enter Python Source Program (Type END to finish):")

inside_comment = False
result = []

while True:
    line = input()

    if line == "END":
        break

    if '"""' in line or "'''" in line:
        inside_comment = not inside_comment
        continue

    if inside_comment:
        continue

    if "#" in line:
        line = line.split("#")[0]

    if line.strip():
        result.append(line)

print("\nOutput:")

for line in result:
    print(line)
```
