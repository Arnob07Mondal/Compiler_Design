**Write a Python Program to Design a DFA to Recognize strings Ending with 01**

```python
string = input("Enter a Binary string: ")
state = 0
for ch in string:
  if ch not in ('0','1'):
    print("Invalid Input")
    exit()
  if state == 0:
    if ch == '0':
      state = 1
    else:
      state = 0
  elif state == 1:
    if ch == '0':
      state = 1
    else:
      state =2
  elif state == 2:
    if ch == '0':
      state = 1
    else:
      state = 0
if state == 2:
  print("Accepted")
else:
  print("Rejected")
```

**Write a python Program to design a DFA to recognize string cotaining even no of 0s and odd no of 1s**

```python
string = input("Enter a Binary string: ")
zeros = 0
ones = 0
for ch in string:
  if ch not in ('0','1'):
    print("Invalid Input")
    exit()
  if ch == '0':
    zeros += 1
  else:
    ones += 1
if zeros % 2 == 0 and ones % 2 == 0:
  print("Accepted")
else:
  print("Rejected")
```
