**Write a python program to design a DFA , that recognizes valid identifire (C keyword)**

```python
identifier = input("Enter an identifier: ")
state = "q0"
for ch in identifier:
  if state=="q0":
    if ch.isalpha() or ch =="_":
      state="q1"
    else:
      state="q2"
  elif state=="q1":
    if ch.isalpha() or ch.isdigit() or ch=="_":
      state="q1"
    else:
      state="q2"
  else:
    state="q2"
if state=="q1":
  print("Valid identifier")
else:
  print("Invalid identifier")
```

**Write a python program to implement a DFA that recognizes integrer and floating point constant**

```python
number = input("Enter a number: ")
state = "q0"
for ch in number:
  if state=="q0":
    if ch.isdigit():
      state="q1"
    else:
      state = "q4"
  elif state == "q1":
    if ch.isdigit():
      state = "q1"
    elif ch == ".":
      state = "q2"
    else:
      state = "q4"
  elif state == "q2":
    if ch.isdigit():
      state = "q3"
    else:
      state = "q4"
  elif state == "q3":
    if ch.isdigit():
      state = "q3"
    else:
      state = "q4"
  elif state == "q4":
    break

if state == "q1" or state == "q3":
  print("Accepted")
else:
  print("Not Accepted")
```

**WAP IN PYTHON TO IMPLEMENT A DFA THAT RECOGNIZES DELIMETERS AND OPERATORS**

```python
token = input("Enter an operator or delimiter: ")

state = "q0"

for ch in token:

    if state == "q0":

        if ch in "+-*/%":
            state = "q1"

        elif ch in "(){}[],;":
            state = "q2"

        elif ch in "=<>!":
            state = "q3"

        elif ch in "&|":
            state = "q6" if ch == "&" else "q7"

        else:
            state = "q5"

    elif state == "q1":

        if ch in "=+-":
            state = "q4"      # +=, -=, *=, /=, %=, ++, --

        else:
            state = "q5"

    elif state == "q3":

        if ch == "=":
            state = "q4"      # ==, !=, <=, >=

        else:
            state = "q5"

    elif state in ["q6", "q7"]:

        if (state == "q6" and ch == "&") or (state == "q7" and ch == "|"):
            state = "q4"      # &&, ||

        else:
            state = "q5"

    else:
        state = "q5"

if state == "q2":
    print("Accepted : Delimiter")

elif state in ["q1", "q3", "q4"]:
    print("Accepted : Operator")

else:
    print("Rejected")
```
