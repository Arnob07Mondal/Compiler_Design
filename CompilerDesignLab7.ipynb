**Write a Python Program to  accept a string using recursive descent parser**

```python
tokens = []
pos = 0


# Match the current token
def match(token):
    global pos

    if pos < len(tokens) and tokens[pos] == token:
        pos += 1
        return True

    return False


# E → T E'
def E():
    if T():
        return E_prime()
    return False


# E' → + T E' | ε
def E_prime():
    if pos < len(tokens) and tokens[pos] == '+':
        match('+')
        return T() and E_prime()

    return True


# T → F T'
def T():
    if F():
        return T_prime()
    return False


# T' → * F T' | ε
def T_prime():
    if pos < len(tokens) and tokens[pos] == '*':
        match('*')
        return F() and T_prime()

    return True


# F → id | ( E )
def F():
    if pos < len(tokens) and tokens[pos] == 'id':
        match('id')
        return True

    if pos < len(tokens) and tokens[pos] == '(':
        match('(')

        if E() and pos < len(tokens) and tokens[pos] == ')':
            match(')')
            return True

    return False


# ============================================
# Display Grammar
# ============================================

print("=" * 50)
print("        RECURSIVE DESCENT PARSER")
print("=" * 50)

print("\nGrammar:")
print("E  → T E'")
print("E' → + T E' | ε")
print("T  → F T'")
print("T' → * F T' | ε")
print("F  → id | ( E )")

print("\nValid terminals:")
print("id, +, *, (, )")

print("\nExamples of valid input:")
print("id")
print("id+id")
print("id+id*id")
print("(id+id)*id")

print("\nNote:")
print("Use 'id' instead of variable names such as a, b, x, y.")
print("=" * 50)


# ============================================
# User Input
# ============================================

input_string = input("\nEnter the expression: ")


# Convert input into tokens
tokens = input_string.replace("(", " ( ")
tokens = tokens.replace(")", " ) ")
tokens = tokens.replace("+", " + ")
tokens = tokens.replace("*", " * ")
tokens = tokens.split()

# Add end marker
tokens.append("$")

pos = 0


# ============================================
# Start Parsing
# ============================================

if E() and tokens[pos] == "$":
    print("\nResult: String Accepted")
else:
    print("\nResult: String Rejected")
```
