**Write a python Program to implement a shift reducing parser using user input string.
Example:- E->E+E | E->E*E | E->id**

```python
stack = []
print("=" * 60)
print(" SHIFT-REDUCE PARSER")
print("=" * 60)
print("\nGrammar:")
print("E -> E + E")
print("E -> E * E")
print("E -> id")
# ============================================
# PRINT FUNCTION
# ============================================
def print_step(stack, tokens, index, action):
    print(
        f"{''.join(stack):<20}"
        f"{''.join(tokens[index:]):<20}"
        f"{action}"
    )
# ============================================
# USER INPUT
# ============================================
expression = input("\nEnter expression: ")
expression = expression.replace(" ", "")
# ============================================
# TOKENIZE INPUT
# ============================================
tokens = []
i = 0
while i < len(expression):
    if expression[i:i+2] == "id":
        tokens.append("id")
        i += 2
    elif expression[i] in ['+', '*']:
        tokens.append(expression[i])
        i += 1
    else:
        print("\nInvalid input!")
        print("String Rejected")
        exit()
# ============================================
# PARSING TABLE
# ============================================
print("\n")
print(f"{'Stack':<20}{'Input':<20}Action")
print("-" * 60)
# ============================================
# SHIFT-REDUCE PROCESS
# ============================================
index = 0
while index < len(tokens):
    # -------------------------
    # SHIFT
    # -------------------------
    stack.append(tokens[index])
    index += 1
    print_step(
        stack,
        tokens,
        index,
        "Shift"
    )
    # -------------------------
    # REDUCE E -> id
    # -------------------------
    if stack[-1] == "id":
        stack[-1] = "E"
        print_step(
            stack,
            tokens,
            index,
            "Reduce E -> id"
        )
    # -------------------------
    # REDUCE E -> E + E
    # REDUCE E -> E * E
    # -------------------------
    reduced = True
    while reduced:
        reduced = False
        if len(stack) >= 3:
            if (
                stack[-3] == "E"
                and stack[-2] in ['+', '*']
                and stack[-1] == "E"
            ):
                operator = stack[-2]
                stack = stack[:-3]
                stack.append("E")
                print_step(
                    stack,
                    tokens,
                    index,
                    f"Reduce E -> E {operator} E"
                )
                reduced = True
# ============================================
# FINAL RESULT
# ============================================
print("-" * 60)
if stack == ["E"]:
    print("String Accepted")
else:
    print("String Rejected")
```
